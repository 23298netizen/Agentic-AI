```python
def classify_email(email_text):
    spam_keywords = ["free", "winner", "urgent", "click here", "prize"]

    text = email_text.lower()
    count = 0

    for keyword in spam_keywords:
        if keyword in text:
            count += 1

    threshold = 2

    if count >= threshold:
        label = "Spam"
    else:
        label = "Legitimate"

    return label, count


if __name__ == "__main__":
    test_emails = [
        ("Congratulations! You are a winner. Claim your free prize now!", "Spam"),
        ("Urgent! Click here to claim your prize.", "Spam"),
        ("Get free offers today. Click here for your prize.", "Spam"),
        ("Hi, can we meet tomorrow to discuss the project?", "Legitimate"),
        ("Your assignment is due on Friday. Please submit it on time.", "Legitimate"),
        ("Thank you for your email. I will reply after checking the document.", "Legitimate")
    ]

    correct = 0

    print("SPAM CLASSIFIER RESULTS")
    print("=" * 60)

    for email, correct_label in test_emails:
        predicted_label, keyword_count = classify_email(email)

        if predicted_label == correct_label:
            correct += 1

        print(f"Email: {email}")
        print(f"Spam keyword count: {keyword_count}")
        print(f"Predicted label: {predicted_label}")
        print(f"Correct label: {correct_label}")
        print("-" * 60)

    accuracy = (correct / len(test_emails)) * 100

    print(f"Correct predictions: {correct}/{len(test_emails)}")
    print(f"Accuracy: {accuracy:.2f}%")
```

```python
def classify_email(email_text, threshold):
    spam_keywords = ["free", "winner", "urgent", "click here", "prize"]

    text = email_text.lower()
    count = 0

    for keyword in spam_keywords:
        if keyword in text:
            count += 1

    if count >= threshold:
        return "Spam"
    else:
        return "Legitimate"


if __name__ == "__main__":
    test_emails = [
        ("Congratulations! You are a winner. Claim your free prize now!", "Spam"),
        ("Urgent! Click here to claim your prize.", "Spam"),
        ("Get free offers today. Click here for your prize.", "Spam"),
        ("Hi, can we meet tomorrow to discuss the project?", "Legitimate"),
        ("Your assignment is due on Friday. Please submit it on time.", "Legitimate"),
        ("Thank you for your email. I will reply after checking the document.", "Legitimate")
    ]

    print("THRESHOLD ACCURACY TEST")
    print("=" * 40)

    for threshold in range(1, 6):
        correct = 0

        for email, correct_label in test_emails:
            predicted_label = classify_email(email, threshold)

            if predicted_label == correct_label:
                correct += 1

        accuracy = (correct / len(test_emails)) * 100

        print(f"Threshold = {threshold}: Accuracy = {accuracy:.2f}%")

    print("=" * 40)
```
